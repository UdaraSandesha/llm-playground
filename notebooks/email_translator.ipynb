```python
# imports

import os
from dotenv import load_dotenv
from IPython.display import Markdown, display
from openai import OpenAI

from llm_playground.ai_provider import AIProvider

GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"
OLLAMA_BASE_URL = "http://localhost:11434/v1"

provider = AIProvider.GEMINI  # Choose OPENAI, GEMINI, or OLLAMA
```

```python
# Load environment variables in a file called .env
load_dotenv(override=True)

if provider is AIProvider.OPENAI:
    api_key_name = "OPENAI_API_KEY"
elif provider is AIProvider.GEMINI:
    api_key_name = "GOOGLE_API_KEY"
elif provider is AIProvider.OLLAMA:
    api_key_name = None
else:
    raise ValueError(f"Unsupported AI provider: {provider}")

if api_key_name is None:
    api_key = None
    print(f"Using local {provider.value} provider; no API key required")
else:
    api_key = os.getenv(api_key_name)
    if not api_key:
        raise ValueError(f"{api_key_name} was not found in the environment")
    if api_key.strip() != api_key:
        raise ValueError(f"{api_key_name} contains leading or trailing whitespace")
    print(f"API key loaded for {provider.value}")

```

```python
system_prompt = """
You are a german professional and official email writing assistant that converts a given email text in English into a professional and official German email text.
Always write the email text in a professional and natural tone, and make sure to use proper German grammar and spelling. The email text should be written in a way that is appropriate for a professional setting, and should convey the same message as the original English text.
Avoid using overly complex or technical language, and instead focus on clear and concise communication. The email text should be easy to read and understand, while still maintaining a professional tone.
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown and use spacing and line breaks as needed.
"""
```

```python
user_prompt = """
Please translate the following email text from English to German

"""
```

```python
email_text = """
Dear Sir/Madam,

My wife is still waiting for her visa approval.
I have already moved into the room and will be paying both of our rent.

Best regards,
Asuramuni Udara Sandesha De Silva
"""
```

```python
messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt + email_text}
]
```

```python
if provider is AIProvider.OPENAI:
    client = OpenAI(api_key=api_key)
    model = "gpt-5.6-terra"
elif provider is AIProvider.GEMINI:
    client = OpenAI(base_url=GEMINI_BASE_URL, api_key=api_key)
    model = "gemini-3.6-flash"
elif provider is AIProvider.OLLAMA:
    client = OpenAI(base_url=OLLAMA_BASE_URL, api_key="ollama")
    model = "llama3.2"
else:
    raise ValueError(f"Unsupported AI provider: {provider}")

response = client.chat.completions.create(model=model, messages=messages)
content = response.choices[0].message.content
display(Markdown(content))
```
